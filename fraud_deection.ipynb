```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("creditcard.csv")

print("Dataset shape:",df.shape)
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
# seprating Fraud and non Fraud transactiom
fraud = df[df['Class'] ==1]
non_fraud = df[df['Class'] == 0]

print("Number of fradulent transaction:", len(fraud))
print("Number of non-fradulent transaction:", len(non_fraud))

# 1.Graph of Transaction Amount Distributtion

plt.figure(figsize=(10,5))
sns.histplot(non_fraud['Amount'], bins=50, color='green', label='Non_Fraud', alpha=0.6)
sns.histplot(fraud['Amount'], bins=50, color='red', label='Fraud', alpha=0.6)
plt.yscale('log')
plt.legend()
plt.title('Transaction Amount Distribution')
plt.xlabel('Amount')
plt.ylabel('Count')
plt.show()

# 2.Graph of Transaction Time Disrtibution

plt.figure(figsize=(10,5))
sns.histplot(non_fraud['Time'], bins=50, color='blue', label='Non_Fraud', alpha=0.6)
sns.histplot(fraud['Time'], bins=50, color='orange', label='Fraud', alpha=0.6)
plt.yscale('log')
plt.legend()
plt.title('Transaction Time Distribution')
plt.xlabel('time (seconds)')
plt.ylabel('Count')
plt.show()

# 3.Corelation Heatmap

plt.figure(figsize=(15,8))
corr = df.corr()
sns.heatmap(corr, cmap='coolwarm', center=0)
plt.title('Corelation Heatmap')
plt.show()

# 4.Top Corelated Features With Fraud
corr_target = corr['Class'].sort_values(ascending=False)
print("Top corelation with Class (Fraud):")
print(corr_target.head(10))
```

Data Preprocessing


```python
x = df.drop('Class', axis = 1)
y = df['Class']

print("Feature Data Shape:", x.shape)
print("Target Data Shape:", y.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
#scaling only 'Time' and 'Amount' columns
x[['Time','Amount']] = scaler.fit_transform(x[['Time','Amount']])
x[['Time','Amount']].head()
```

```python
# Handling Imbalanced Data ith SMOTE

from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state = 42)
x_resampled, y_resampled = smote.fit_resample(x,y)

print("Before smote:", y.value_counts())
print("After  smote:", y_resampled.value_counts())
      
```

```python
# Train Test split
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x_resampled, y_resampled, test_size=0.2, stratify=y_resampled)

print("Training set shape:", x_train.shape)
print("Test set shape:", x_test.shape)
```

Model Building & Evaluation

```python
#1. Logistic Regression (Baseline Model)
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# training the mmodel
log_reg = LogisticRegression(max_iter=1000, random_state=42)
log_reg.fit(x_train, y_train)

#Predictions
y_pred_lr = log_reg.predict(x_test)

#evaluation
print(" Logistic Regression Performance:\n")
print(classification_report(y_test, y_pred_lr))
print("ROC-AUC Score:", roc_auc_score(y_test,y_pred_lr))

```

```python
#2. Random Forest Classifier

from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(x_train, y_train)

y_pred_rf = rf.predict(x_test)

print(" Random Forest Performance:\n")
print(classification_report(y_test, y_pred_lr))
print("ROC-AUC Score:", roc_auc_score(y_test,y_pred_lr))
```

```python
#3. XGBoost Classifier (Advanced Model)

from xgboost import XGBClassifier

xgb = XGBClassifier(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=6,
    random_state=42,
    use_lable_encoder=False,
    eval_matric='Logoss'
)
xgb.fit(x_train, y_train)

y_pred_xgb = xgb.predict(x_test)

print("XGBoost Performance:\n")
print(classification_report(y_test, y_pred_xgb))
print("ROC-AUC Score:", roc_auc_score(y_test, y_pred_xgb))
```

```python
# Visual Evaluation — Confusion Matrix
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import ConfusionMatrixDisplay

models={
    "Logistic Reggerision": y_pred_lr,
    "Random Forest": y_pred_rf,
    "XGBoost": y_pred_xgb
}

plt.figure(figsize=(15, 4))

for i, (model, preds) in enumerate(models.items()):
    plt.subplot(1,3, i + 1)
    ConfusionMatrixDisplay.from_predictions(y_test, preds, cmap="Blues", ax=plt.gca(), colorbar=False)
    plt.title(model)

plt.tight_layout()
plt.show()
```

```python
# Comparing Model Metrics Side-by-Side
from sklearn.metrics import precision_score, recall_score, f1_score

results= pd.DataFrame({
    "Model": ["Logistic Regression","Random Forest", "XGBoost"],
    "Precision": [
        precision_score(y_test, y_pred_lr),
        precision_score(y_test, y_pred_rf),
        precision_score(y_test, y_pred_xgb)
    ],
    "F1-Score": [
        f1_score(y_test, y_pred_lr),
        f1_score(y_test, y_pred_rf),
        f1_score(y_test, y_pred_xgb)
    ],
    "Recall": [
        recall_score(y_test, y_pred_lr),
        recall_score(y_test, y_pred_rf),
        recall_score(y_test, y_pred_xgb)
    ],
    "ROC-AUC": [
        roc_auc_score(y_test, y_pred_lr),
        roc_auc_score(y_test, y_pred_rf),
        roc_auc_score(y_test, y_pred_xgb)
    ]
})

results
        
```

```python
# Feature Importance( For Tree Models)
importances = rf.feature_importances_
indices = np.argsort(importances)[-10:]  #Gives Top 10 Important Features

plt.figure(figsize=(8,5))
plt.title("Top 10 Important Features (Random Forest}")
sns.barplot(x=importances[indices], y=x_train.columns[indices])
plt.show()
```

```python
best_model_name = results.sort_values(by='F1-Score', ascending=False).iloc[0]['Model']
print("Best Model based on F1-Score:", best_model_name)
```

Saving the Final Model & Making Predictions

```python
import joblib

# Saving Random Forest
joblib.dump(rf, 'random_forest_model.pkl')
print("Random Forest model  saved successfully as 'random_forest_model.pkl'")

# Saving XGBoost
joblib.dump(xgb, 'xgboost_model.pkl')
print("XGBoost model  saved successfully as 'xgboost_model.pkl'")
```

```python
# Saving the scaler for new data
joblib.dump(scaler, 'scaler.pkl')
print("Scaler saved successfully as 'scaler.pkl'")
```
